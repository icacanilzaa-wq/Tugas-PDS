```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


#Data
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei']
penjualan = [150, 200, 180, 250, 125]

#6x4 inci adalah ukuran yg kecil dan tidak full screen
plt.figure(figsize=(8, 5))

plt.bar(bulan, penjualan, color='blue')
plt.title('penjualan per bulan')
plt.xlabel('bulan')
plt.ylabel('jumlah')

#color='red'
plt.show()

x = np.linspace(0, 10, 100)
y = np.cos(x)

plt.plot(x, y)
plt.title("Contoh grafik dengan style")
plt.show()
```

```python
#Data
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Juni', 'Juli']
penjualan = [150, 200, 180, 250, 125, 220, 130]

plt.figure(figsize=(6, 4))

#Membuat grafik dengan palette warna yang lebih modern
sns.barplot(x=bulan, y=penjualan, hue=bulan, palette="colorblind", legend=False)

plt.title('penjualan per bulan')
plt.xlabel('bulan')
plt.ylabel('jumlah')

plt.show()
```
