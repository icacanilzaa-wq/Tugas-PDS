```python
# T2.2: Mengecek Bentuk (Shape) 
a = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print(a.shape) 

# T2.3: Array 1D Sederhana
a = np.array([5, 10, 15, 20])
print(a)

# T2.4: Mengubah Bentuk Array 
a = np.array([[2, 4, 6, 8], [10, 12, 14, 16]])
b = a.reshape(4, 2)  # Mengubah dari 2x4 menjadi 4x2
print(b)

# T2.5: Reshape ke Array 3D
a = np.arange(18)  # Array 0 sampai 17
b = a.reshape(2, 3, 3)  # 2 blok, 3 baris, 3 kolom
print(b)

# T3.2 & T3.3: Menggunakan np.arange()
print(np.arange(5))
print(np.arange(5.0))
print(np.arange(4, 9))
print(np.arange(3, 15, 4))
print(np.arange(1, 8, 2))


# T4.2: Menggunakan np.linspace()
urutan = np.linspace(0, 20, 5, dtype=int)
print(urutan) 

# T9: Pembagian Element-wise Matriks
A = np.array([[15, 30], [45, 60]])
B = np.array([[3, 5], [9, 10]])
C = A / B
print(C)

# T10.19: Penggabungan Array dengan np.append()
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])
hasil = np.append(arr1, arr2)
print(hasil)

# T10.20: Menyisipkan Nilai dengan np.insert()
arr1 = np.array([100, 200, 300, 400, 500])

# Menyisipkan nilai [99, 99] pada indeks ke-3
arr1_baru = np.insert(arr1, 3, [99, 99])
print(arr1)
print(arr1_baru)

# Data interval nilai baru
interval = np.array([50, 60, 70, 80, 90, 100]) 
frekuensi = np.array([4, 8, 15, 10, 3])

# Menghitung titik tengah 
midpoints = (interval[:-1] + interval[1:]) / 2

# Menghitung Mean, Varians, dan Standar Deviasi
mean = np.sum(midpoints * frekuensi) / np.sum(frekuensi)
varians = np.sum(((midpoints - mean) ** 2) * frekuensi) / np.sum(frekuensi)
std_deviation = np.sqrt(varians)

print(round(mean, 2))
print(round(varians, 2))
print(round(std_deviation, 2))
```

PANDAS

```python
import numpy as np
import pandas as pd

# 1. Menampilkan Pandas Series (1-Dimensi)
ds = pd.Series([2, 4, 6, 8, 10])
print("Pandas Series:")
print(ds)

# 2. Membuat Series dari NumPy Array
np1 = np.array(['x', 'y', 'z'])
data_np = pd.Series(np1)
print("\nSeries dari NumPy:")
print(data_np)

# Custom Index
data_np_custom = pd.Series(np1, index=[2, 3, 4])
print("\nSeries dengan Custom Index:")
print(data_np_custom)

# 3. Operasi Aritmatika pada Series
ds1 = pd.Series([4, 6, 8, 10, 12])
ds2 = pd.Series([3, 5, 7, 9, 11])

print("\nPenjumlahan:", ds1 + ds2)
print("Pengurangan:", ds1 - ds2)
print("Perkalian  :", ds1 * ds2)
print("Pembagian  :", ds1 / ds2)

# 4. Konversi Pandas Series ke Python List
print("\nKonversi Series ke List:", ds1.tolist())

# 5. Contoh DataFrame
data = {
    "produk": ["Laptop", "HP", "Printer", "Tablet", "Kamera"],
    "kategori": ["A", "B", "C", "D", "E"],
    "harga": [10000, 5000, 3000, 7000, 8000]
}
df = pd.DataFrame(data)
print("\nDataFrame:")
print(df)
```
