NUMPY

```python
import numpy as np

# Membuat array 1-Dimensi
nilai = np.array([85, 55, 40, 90])

# Mengakses data pada array berdasarkan indeks
print("Elemen indeks ke-3:", nilai[3])  # Output: 90

# Pengaksesan & Slicing Array
data = np.array([1, 2, 3, 4, 5])
print(data[1])       
print(data[0:2])    
print(data[1:])      
print(data[-2:])     

# Menampilkan bentuk (shape) Array 2-Dimensi
a_2d = np.array([[1, 2, 3, 4], [5, 66, 7, 8]])
print(a_2d.shape)       
```

```python
# Reshape Matriks 2D (dari 2x3 menjadi 3x2)
a = np.array([[1, 2, 3], [4, 5, 6]])
b = a.reshape(3, 2)
print(b)

# Reshape ke Array 3D (2x4x3)
a_seq = np.arange(24)  # Membuat deret angka 0-23
b_3d = a_seq.reshape(2, 4, 3)
print(b_3d)
```

```python
# Penggunaan np.arange()
print(np.arange(3))       
print(np.arange(3.0))      
print(np.arange(3, 7))    
print(np.arange(2, 12, 3)) 
print(np.arange(2, 6, 5)) 
print(np.arange(3, 10, 3)) 

# Penggunaan np.linspace() (membuat deret terbagi rata)
urutan = np.linspace(0, 10, 4, dtype=int)
print(urutan) 
```

```python
# Penggabungan dua array dengan np.append()
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
hasil_append = np.append(arr1, arr2)
print(hasil_append)  

# Menyisipkan nilai dengan np.insert()
arr1_base = np.array([1, 2, 3, 4, 5])
# Menyisipkan nilai [9, 9, 9] pada indeks ke-2
arr1_baru = np.insert(arr1_base, 2, [9, 9, 9])

print("Arr1 Asli:", arr1_base)
print("Arr1 Baru:", arr1_baru)
```

```python
# Data interval & frekuensi
interval = np.array([20, 30, 40, 50, 60, 70]) 
frekuensi = np.array([5, 12, 18, 10, 7])      

# Menhitung midpoint (titik tengah interval)
midpoints = (interval[:-1] + interval[1:]) / 2

# Menghitung Mean, Varians, dan Standar Deviasi
mean = np.sum(midpoints * frekuensi) / np.sum(frekuensi)
varians = np.sum(((midpoints - mean) ** 2) * frekuensi) / np.sum(frekuensi)
std_deviation = np.sqrt(varians)

print("Mean              :", mean)
print("Varians           :", varians)
print("Standard Deviation:", std_deviation)
```

PANDAS

```python
import numpy as np
import pandas as pd

# 1. Menampilkan Pandas Series (1-Dimensi)
ds = pd.Series([2, 4, 6, 8, 10])
print("Pandas Series:")
print(ds)

# 2. Membuat Series dari NumPy Array
np1 = np.array(['x', 'y', 'z'])
data_np = pd.Series(np1)
print("\nSeries dari NumPy:")
print(data_np)

# Custom Index
data_np_custom = pd.Series(np1, index=[2, 3, 4])
print("\nSeries dengan Custom Index:")
print(data_np_custom)

# 3. Operasi Aritmatika pada Series
ds1 = pd.Series([4, 6, 8, 10, 12])
ds2 = pd.Series([3, 5, 7, 9, 11])

print("\nPenjumlahan:", ds1 + ds2)
print("Pengurangan:", ds1 - ds2)
print("Perkalian  :", ds1 * ds2)
print("Pembagian  :", ds1 / ds2)

# 4. Konversi Pandas Series ke Python List
print("\nKonversi Series ke List:", ds1.tolist())

# 5. Contoh DataFrame
data = {
    "produk": ["Laptop", "HP", "Printer", "Tablet", "Kamera"],
    "kategori": ["A", "B", "C", "D", "E"],
    "harga": [10000, 5000, 3000, 7000, 8000]
}
df = pd.DataFrame(data)
print("\nDataFrame:")
print(df)
```
